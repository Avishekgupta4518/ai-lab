```python

# Step 2: Initialize translator
translator = Translator()
```

```python
# Step 3: Translate English text to Nepali
result = translator.translate("How are you?", src='en', dest='ne')
```

```python
# Step 1: Import translator
from googletrans import Translator

# Step 2: Initialize translator
translator = Translator()

# Step 3: Translate English text to Nepali (no await needed)
result = translator.translate("How are you?", src='en', dest='ne')

# Step 4: Print translated text
print("Nepali Translation:", result.text)
```
